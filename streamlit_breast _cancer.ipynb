```python
from sklearn.model_selection import train_test_split
```

```python
full, test, y_full, y_test = train_test_split(df, y, test_size = 0.2, random_state= 3)
train, val, y_train, y_val = train_test_split(full, y_full, test_size = 0.2, random_state = 3)
```

```python
full.shape
```

```python
y_full.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, recall_score, precision_score
```

```python
## Logistic Regression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(train, y_train)
```

```python
lr_pred = lr.predict(val)
```

```python
accuracy_score(y_val, lr_pred)
```

```python
confusion_matrix(y_val, lr_pred)
```

```python
recall_score(y_val, lr_pred)
```

```python
precision_score(y_val, lr_pred)
```

        DECISION TREE

```python
dt = DecisionTreeClassifier(max_depth = 21)
```

```python
dt.fit(train, y_train)
```

```python
dt_pred = dt.predict(val)
```

```python
accuracy_score(y_val, dt_pred)
```

```python
confusion_matrix(y_val, dt_pred)
```

```python
precision_score(y_val, dt_pred)
```

```python
recall_score(y_val, dt_pred)
```
